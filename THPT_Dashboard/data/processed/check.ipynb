```python
import pandas as pd
import glob

files = sorted(glob.glob("thpt202*.csv"))

# Danh sách tất cả môn
subjects = [
    'toan','ngu_van','ngoai_ngu',
    'vat_ly','hoa_hoc','sinh_hoc',
    'lich_su','dia_ly','gdcd'
]

for file in files:
    df = pd.read_csv(file)

    # Đếm số môn có điểm
    subject_count = df[subjects].notna().sum(axis=1)

    # Những người < 3 môn
    mask_less_3 = subject_count < 3
    df_less_3 = df[mask_less_3]

    print(f"\n📂 {file}")
    print(f"Tổng thí sinh: {len(df):,}")
    print(f"❌ < 3 môn: {len(df_less_3):,} ({len(df_less_3)/len(df)*100:.2f}%)")
    print(f"✔️ ≥ 3 môn: {len(df) - len(df_less_3):,}")
```

```python
import pandas as pd

# ==========================================
# 1. Load 2 file
# ==========================================
df_06 = pd.read_csv("thpt2025_06.csv")
df_18 = pd.read_csv("thpt2025_18.csv")

# ==========================================
# 2. Cột chung
# ==========================================
common_cols = [
    'sbd','toan','ngu_van','ngoai_ngu',
    'vat_ly','hoa_hoc','sinh_hoc',
    'lich_su','dia_ly','gdcd'
]

df_06 = df_06[common_cols]
df_18 = df_18[common_cols]

# ==========================================
# 3. Gộp dữ liệu
# ==========================================
df_2025 = pd.concat([df_06, df_18], ignore_index=True)

# ==========================================
# 4. Ép kiểu sbd để sort chuẩn (quan trọng)
# ==========================================
df_2025['sbd'] = pd.to_numeric(df_2025['sbd'], errors='coerce')

# ==========================================
# 5. Sort theo sbd
# ==========================================
df_2025 = df_2025.sort_values('sbd').reset_index(drop=True)

# ==========================================
# 6. Lưu file
# ==========================================
df_2025.to_csv("thpt2025.csv", index=False, encoding="utf-8-sig")

print("Done! Saved as thpt2025.csv")
print(df_2025.shape)
df_2025.head()
```

```python
import pandas as pd
import glob

# danh sách các cột môn học
subject_cols = [
    'toan','ngu_van','ngoai_ngu',
    'vat_ly','hoa_hoc','sinh_hoc',
    'lich_su','dia_ly','gdcd'
]

files = glob.glob("thpt*.csv")

for f in files:
    df = pd.read_csv(f)
    
    before = len(df)

    # loại bỏ dòng mà TẤT CẢ môn đều NULL
    df_clean = df[~df[subject_cols].isna().all(axis=1)]

    after = len(df_clean)
    
    # lưu lại file (ghi đè)
    df_clean.to_csv(f, index=False)

    print(f"\n📂 {f}")
    print(f"👉 Trước: {before:,}")
    print(f"👉 Sau:   {after:,}")
    print(f"🧹 Đã xóa: {before - after:,}")
```
